| Đoạn | Loại lỗi | Nguyên nhân | Cách sửa |
| :---: | :--- | :--- | :--- |

```python
# 1. Nhập tên khách hàng và giá của 4 sản phẩm (số thực)
ten_kh = input('Nhập tên khách hàng: ')
g1 = float(input('Giá sản phẩm 1: '))
g2 = float(input('Giá sản phẩm 2: '))
g3 = float(input('Giá sản phẩm 3: '))
g4 = float(input('Giá sản phẩm 4: '))
gia = [g1, g2, g3, g4]

# 2. Tính tiền hàng, giảm giá 5%, VAT 8% và tổng thanh toán
tong_tien = sum(gia)
giam_gia = tong_tien * 5 / 100
tien_sau_giam = tong_tien - giam_gia
vat = tien_sau_giam * 8 / 100
tong_thanh_toan = tien_sau_giam + vat

# 3. Tìm sản phẩm đắt nhất, rẻ nhất và giá trung bình
gia_max = max(gia)
gia_min = min(gia)
gia_tb = tong_tien / len(gia)

# 4. Tạo bản sao danh sách giá và sắp xếp giảm dần
gia_sx = gia.copy()
gia_sx.sort(reverse=True)

# 5. Kiểm tra miễn phí vận chuyển (từ 250.000 trở lên)
mien_phi_van_chuyen = tong_thanh_toan >= 250000

# 6. In hóa đơn đóng khung đẹp
print('=' * 40)
print('HÓA ĐƠN SIÊU THỊ')
print('=' * 40)
print('Khách hàng:', ten_kh)
print('Giá sau sắp xếp giảm dần:', gia_sx)
print('Tổng tiền hàng:', tong_tien)
print('Giảm giá 5%:', giam_gia)
print('Tiền sau giảm:', tien_sau_giam)
print('VAT 8%:', vat)
print('Tổng thanh toán:', tong_thanh_toan)
print('Đắt nhất:', gia_max, '- Rẻ nhất:', gia_min)
print('Giá trung bình:', gia_tb)
print('Miễn phí vận chuyển:', mien_phi_van_chuyen)
print('=' * 40)
```
